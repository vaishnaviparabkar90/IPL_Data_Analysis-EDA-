# **IPL Data Analysis**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
#loading the datasets

matches_df=pd.read_csv("/content/IPL Matches 2008-2020.csv")
balls_df=pd.read_csv("/content/IPL Ball-by-Ball 2008-2020.csv")
```

```python
matches_df.head(2)
```

```python
balls_df.head(2)
```

```python
#1. What was the count of matches played in each season?

matches_df['date']=pd.to_datetime(matches_df['date'])
matches_df['season']=matches_df['date'].dt.year

matches_df['season'].value_counts().sort_index()
```

```python
#2. How many runs were scored in each season?

merged_df = pd.merge(balls_df, matches_df[["id", "season"]], on="id")
merged_df.groupby('season').total_runs.sum()

balls_df
```

```python
#3. What were the runs scored per match in different seasons?

merged_df.groupby(['season','id']).total_runs.sum()
```

```python
#4. Who has umpired the most?

#matches_df['umpire1'].value_counts().idxmax()

pd.concat([matches_df["umpire1"], matches_df["umpire2"]]).value_counts().idxmax()
```

```python
#5. Which team has won the most tosses?

matches_df['toss_winner'].value_counts().idxmax()
```

```python
#6. What does the team decide after winning the toss?

matches_df['toss_decision'].value_counts()
```

```python
#7. How does the toss decision vary across seasons?

matches_df.groupby(['season','toss_decision']).size()
```

```python
#8. Does winning the toss imply winning the game?

(matches_df["toss_winner"] == matches_df["winner"]).value_counts()
```

```python
#9. How many times has the chasing team won the match?

matches_df["chasing_team"] = np.where(
    matches_df["toss_decision"] == "field",
    matches_df["toss_winner"],
    np.where(
        matches_df["team1"] == matches_df["toss_winner"],
        matches_df["team2"],
        matches_df["team1"]
    )
)

len(matches_df[matches_df["chasing_team"] == matches_df["winner"]])
```

```python
len(matches_df[matches_df["result"] == "wickets"])
```

```python
#10. Which all teams had won this tournament?

matches_df['winner'].unique()
```

```python
#11. Which team has played the most number of matches?

pd.concat([matches_df['team1'], matches_df['team2']]).value_counts().idxmax()
```

```python
#12. Which team has won the most number of times?

matches_df['winner'].value_counts().idxmax()
```

```python
#13. Which team has the highest winning percentage?



(
    matches_df['winner'].value_counts()
    / pd.concat([matches_df['team1'], matches_df['team2']]).value_counts()
    * 100
).sort_values(ascending=False)
```

```python
#14. Is there any lucky venue for a particular team?

matches_df.groupby(['venue', 'winner']).size().sort_values(ascending=False)
```

```python
#15. Innings wise comparison between teams

balls_df.groupby(['batting_team','inning']).total_runs.sum().unstack()
```

```python
#16. Which team has scored the most number of 200+ scores?

team_scores=balls_df.groupby(['batting_team','id']).total_runs.sum()
team_scores[team_scores >= 200].groupby('batting_team').count().idxmax()
```

```python
#17. Which team has conceded 200+ scores the most?

team_scores = balls_df.groupby(['id', 'batting_team', 'bowling_team'])['total_runs'].sum()
team_scores[team_scores >= 200].groupby('bowling_team').count().idxmax()
```

```python
#18. What was the highest run scored by a team in a single match?
balls_df.groupby(['id', 'batting_team'])['total_runs'].sum().idxmax()
```

```python
#19. Which is the biggest win in terms of run margin?

matches_df[matches_df['result'] == 'runs']['result_margin'].max()
```

```python
#20.Which batsmen have played the most number of balls?

balls_df['batsman'].value_counts().idxmax()
```

```python
#21. Who are the leading run-scorers of all time?

balls_df.groupby('batsman')['batsman_runs'].sum().idxmax()
```

```python
#22. Who has hit the most number of 4's?

balls_df[
    (balls_df['batsman_runs'] == 4) &
    (balls_df['non_boundary'] == 0)
].groupby('batsman').size().idxmax()
```

```python
#23. Who has hit the most number of 6's?

balls_df[
    (balls_df['batsman_runs'] == 6) &
    (balls_df['non_boundary'] == 0)
].groupby('batsman').size().idxmax()
```

```python

#24. Who has the highest strike rate?

runs = balls_df.groupby('batsman')['batsman_runs'].sum()

balls_faced = balls_df[balls_df['extras_type'] != 'wides'].groupby('batsman').size()

strike_rate = (runs / balls_faced) * 100
strike_rate.idxmax()
```

```python
#25. Who is the leading wicket-taker?

balls_df[balls_df['is_wicket']==1].groupby('bowler').size().idxmax()
```

```python
#26. Which stadium has hosted the most number of matches?

matches_df['venue'].value_counts().idxmax()
```

```python
#27. Who has won the most MOM awards?

matches_df['player_of_match'].value_counts().idxmax()
```

```python
#28. What is the count of fours hit in each season?

merged_df[
    (merged_df['batsman_runs'] == 4) &
    (merged_df['non_boundary'] == 0)
].groupby('season').size()
```

```python
#29. What is the count of sixes hit in each season?

merged_df[
    (merged_df['batsman_runs'] == 6) &
    (merged_df['non_boundary'] == 0)
].groupby('season').size()
```

```python
#30. What is the count of runs scored from boundaries in each season?


merged_df[
    (merged_df['batsman_runs'].isin([4, 6])) &
    (merged_df['non_boundary'] == 0)
].groupby('season')['batsman_runs'].sum()
```

```python
#31. What is the run contribution from boundaries in each season?


boundary_runs = merged_df[
    (merged_df['batsman_runs'].isin([4, 6])) &
    (merged_df['non_boundary'] == 0)
].groupby('season')['batsman_runs'].sum()

total_runs = merged_df.groupby('season')['total_runs'].sum()

run_contribution = (boundary_runs / total_runs) * 100

run_contribution.round(2)
```

```python
#32. Which team has scored the most runs in the first 6 overs?


first_6_overs = balls_df[balls_df['over'] < 6]

first_6_overs.groupby('batting_team')['total_runs'].sum().sort_values(ascending=False)
```

```python
#33. Which team has scored the most runs in the last 4 overs?

last_4_overs = balls_df[balls_df['over'] >= 16]

last_4_overs.groupby('batting_team')['total_runs'].sum().sort_values(ascending=False)
```

```python
#34. Which team has the best scoring run-rate in the first 6 overs?

first_6_overs = balls_df[balls_df['over'] < 6]

run_rate = (
    first_6_overs.groupby('batting_team')['total_runs'].sum() / 6
).idxmax()

run_rate
```

```python
#35. Which team has the best scoring run-rate in the last 4 overs?
last_4_overs = balls_df[balls_df['over'] >= 16]

run_rate = (
    last_4_overs.groupby('batting_team')['total_runs'].sum() / 4
).sort_values(ascending=False)

run_rate
```
